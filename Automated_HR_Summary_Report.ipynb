```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

saved_report = pd.read_csv('Automated_HR_Summary_Report.csv')

plt.figure(figsize=(7, 4))
sns.set_theme(style="white")

max_rate = saved_report['Attrition_Rate(%)'].max()
colors = ['#ff4d4d' if val == max_rate else '#4a90e2' for val in saved_report['Attrition_Rate(%)']]

# Bar Chart வரைதல்
chart = sns.barplot(
    data=saved_report,
    x='Department',
    y='Attrition_Rate(%)',
    palette=colors,
    hue='Department',
    legend=False
)


plt.title('Critical Attrition Hotspots by Department', fontsize=14, fontweight='bold', color='#2c3e50', pad=15)
plt.xlabel('Department', fontsize=11, fontweight='bold', color='#34495e')
plt.ylabel('Attrition Rate (%)', fontsize=11, fontweight='bold', color='#34495e')


sns.despine(top=True, right=True)

for p in chart.patches:
    height = p.get_height()
    chart.annotate(f"{height}%", 
                   (p.get_x() + p.get_width() / 2., height), 
                   ha='center', va='bottom', 
                   fontsize=11, fontweight='bold', color='#2c3e50', 
                   xytext=(0, 5), textcoords='offset points')

plt.tight_layout()

plt.savefig('Unique_HR_Attrition_Chart.png', dpi=300)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. டேட்டாவை லோட் செய்தல் (பாதை சரியாக உள்ளதா எனச் சரிபார்க்கவும்)
file_path = r"C:\Users\acer\Downloads\WA_Fn-UseC_-HR-Employee-Attrition.csv"
df = pd.read_csv(file_path)

# 2. சார்ட் அமைப்புகளை உருவாக்குதல்
plt.figure(figsize=(7, 5))
sns.set_theme(style="white")

# 3. OverTime மற்றும் Attrition தொடர்பைக் காட்டும் Count Plot
chart = sns.countplot(
    data=df,
    x='OverTime',
    hue='Attrition',
    palette=['#4a90e2', '#ff4d4d']
)

# 4. தலைப்பு மற்றும் லேபிள்கள்
plt.title('Impact of OverTime on Employee Attrition', fontsize=14, fontweight='bold', color='#2c3e50', pad=15)
plt.xlabel('OverTime (Did they work overtime?)', fontsize=11, fontweight='bold', color='#34495e')
plt.ylabel('Number of Employees', fontsize=11, fontweight='bold', color='#34495e')

# லெஜெண்ட் (Legend) மற்றும் பார்டர்களை மாற்றுவது
plt.legend(title='Attrition', loc='upper right')
sns.despine(top=True, right=True)

plt.tight_layout()

# சார்ட்டை சேமிக்க:
plt.savefig('OverTime_vs_Attrition_Chart.png', dpi=300)

plt.show()
```
